```python
import pandas as pd

# Load your Excel file
file_path = '/Users/hariskhan/Downloads/Census_Data.xlsx'
df = pd.read_excel(file_path)

# List of county names to exclude (in lowercase for case-insensitive matching)
counties_to_exclude = [
    'dallas', 'tarrant', 'denton', 'collin', 
    'rockwall', 'kaufman', 'ellis', 'johnson', 
    'parker', 'wise'
]

# Create a regex pattern that matches any of these counties
pattern = '|'.join(counties_to_exclude)

# Filter out rows where the 'NAME' column (converted to lowercase) contains any of the counties
filtered_df = df[~df['NAME'].str.lower().str.contains(pattern, na=False)]

# Display the first few rows of the filtered DataFrame
print(filtered_df.head())
```
